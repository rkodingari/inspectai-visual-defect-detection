# InspectAI dataset EDA
Run dataset preparation first. Change `DATA_DIR` to `data/processed/mvtec_bottle` for the real dataset.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

DATA_DIR = Path('../data/processed/synthetic')
```

```python
rows = []
for split in ('train', 'val', 'test'):
    for label in ('normal', 'defective'):
        for path in (DATA_DIR / split / label).glob('*'):
            with Image.open(path) as im:
                rows.append({'split': split, 'label': label, 'path': path, 'width': im.width, 'height': im.height})
df = pd.DataFrame(rows)
display(df.head(), df.groupby(['split', 'label']).size().rename('count'), df[['width','height']].describe())
```

```python
df.groupby(['split', 'label']).size().unstack().plot.bar(title='Class balance')
plt.ylabel('Images'); plt.tight_layout()
```

```python
fig, axes = plt.subplots(2, 4, figsize=(12, 6))
for row, label in enumerate(('normal', 'defective')):
    for ax, path in zip(axes[row], df[(df.split == 'train') & (df.label == label)].path.head(4)):
        ax.imshow(Image.open(path)); ax.set_title(label); ax.axis('off')
plt.tight_layout()
```

## Leakage and limitations
Check near-duplicates before a serious experiment. MVTec defect images originate from its official test partition; InspectAI splits those examples to enable supervised binary classification, so these results must not be compared with the canonical unsupervised anomaly-detection benchmark.
